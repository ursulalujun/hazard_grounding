```python
import json
import os
with open('earbench_malicious.json') as f:
    data = json.load(f)

new_data = []
for d in data:
    if d['analysis']['result'] == 'benign':
        continue
    basename = os.path.basename(d['image_path'])
    image_path = os.path.join('data_pipeline/supplement/earbench_rewritten_images', d['scene'], basename)
    if not os.path.exists(os.path.join('/mnt/shared-storage-user/luxiaoya/code/EAI/SafePlanner/risk_grounding', image_path)):
        print('image not found')
        continue
    new_data.append({
        "image_path": image_path,
        "scene_type": d['scene'],
        "safety_risk": {
            "safety_principle": "34. Do not execute malicious instruction",
            "action": d['instruction'],
            "safety_hazard": d['explanation'],
            "edit_image_path": image_path,
            "annotation": {
                "target_object": {},
                "constraint_object": {}
            },
            "cot": '<think> ' + d['analysis']['reason'] + ' </think>' + '[safety_hazard]['+d['explanation']+'], violating [safety_principle][34. Do not execute malicious instruction]',
            "principle_id": 34
        }
    })
print(new_data[0]['safety_risk']['cot'])
print(len(new_data))
```
